
## <span style="color: purple;">🧩 Concept: What is a Notebook?</span>

<div class="alert alert-success">

**A notebook** is an interactive page where you can write and run Python step-by-step.

</div>

- **Markdown cells** show text, headings, and instructions.
- **Code cells** run Python and show results immediately.

<div class="alert alert-info">

**Tip:** After changing a code cell, click ▶ to run it again and see what changes.

</div>


In the code cell below, we use `print()` to show text on the screen.

```python
print("You are running your first code cell.")
print("✓ Code cell ran successfully")
```


- `print()` shows output
- Each line is run in order
- You can edit text and run again to see a change

<div class="alert alert-info">

Python can do arithmetic like a calculator. Try running the next cell.

</div>

```python
# Addition
print("2 + 3 =", 2 + 3)

# Multiplication
print("4 * 5 =", 4 * 5)

# Division
print("10 / 2 =", 10 / 2)

# Power
print("3 ** 2 =", 3 ** 2)

print("✓ Math calculations complete")
```


## <span style="color: purple;">Step 3 - Variables</span>

<div class="alert alert-success">

**A variable** stores a value you can reuse and change.

</div>

Example: `x = 10` stores 10 in `x`.

```python
# Define x before using it
x = 10
print("x =", x)

# Update x
x = x + 5
print("x after adding 5 =", x)

# Another variable
y = x * 2
print("y =", y)

# Combine text and values
print("If x is", x, "and y is", y, ", then x + y is", x + y)

print("✓ Variables used successfully! Good job!")
```
